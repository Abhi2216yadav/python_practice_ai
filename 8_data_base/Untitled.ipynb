```python
import sqlite3
```

```python
conn = sqlite3.connect("shop.db")

cursor = conn.cursor()
print("databse is ready")
```

```python
cursor.execute("""
    CREATE TABLE IF NOT EXISTS reviews (
        id INTEGER PRIMARY KEY,
        text TEXT,
        label TEXT,
        core INTEGER
        )
    """)
```

```python
cursor.execute(
    "INSERT INTO reviews (text, label, core) VALUES (?, ?, ?)",
    ("The biryani was amazing!", "positive", 5)
)
```

```python
conn.commit()
```

```python
more_reviews = [
     ("The biryani was amazing!", "positive", 5),
     ("The Chai was amazing!", "positive", 5),
     ("The pasta was very bad in test!", "negative", 1),
     ("The samosa was oilling!", "negative", 2),
]
cursor.executemany(
    "INSERT INTO reviews (text, label, core) VALUES (?, ?, ?)",
    more_reviews
)

conn.commit()
```
