```python
def Calculate_Margin_and_Profit(revenue, expencess):
    profit = revenue - expencess
    margin = profit * 100 / revenue
    return (profit, margin)
    
```

```python
result = Calculate_Margin_and_Profit(50, 10)
result
```

```python
#type hint
def Calculate_Margin_and_Profit(revenue : float, expencess : float) -> float:
    profit = revenue - expencess
    margin = profit * 100 / revenue
    return (profit, margin)
    
```

```python
result = Calculate_Margin_and_Profit(450, 110)
result
```

```python
def total_expences(rent, phone_bill):
    return rent + phone_bill

total_expences(1200, 239)
```

```python
def total_expences(*args):
    return sum(args)

total_expences(1300, 239)
```

```python
def total_expencesses(**kargs):
    total = 0
    for key, value in kargs.items():
        total += value
    return total

total_expencesses(rent = 1200, phone_bill = 120, waymo = 20)
```

### Chai bill calculator

```python
def Chai_bill(cups, *addons, **discount):
    bill = 15 * cups
    
    # print(bill)

    bill += 5 * len(addons)
    
    for key, value in discount.items():
        bill -= bill * value /100
    return bill

# Chai_bill(4, "masala", discount = 10)
print(Chai_bill(4, "ginger", "masala", discount_percent=10))
```

### OOPS concepts

```python
class Employee:
    def __init__(self, name , salary):
        self.name = name
        self.salary = salary

    def calculate_bonous(self):
        return self.salary * 0.10

emp1 = Employee("Abhay", 800000)
emp2 = Employee("Ram", 700000)

# emp1.name
# emp1.salary
emp1.calculate_bonous()
```

```python
from datetime import date

class Human:
    def __init__(self, name , birth_year):
        self.name = name
        self.birth_year = birth_year
    def get_age(self):
        current_year = date.today().year
        return current_year - self.birth_year
    def intro(self):
        print(f"My name is {self.name}, and i am {self.get_age()} year old")

h1 = Human("Abhay" , 2006)
h1.intro()
h1.get_age()
```

### Creating Youtube channal project

```python
class YoutubeChannel:

    def __init__(self, name):
        self.name = name
        self.count = 0

    def subscribe(self):
        self.count += 1
        print(f"Thanks for subscribing! Total: {self.count}")

    def unsubscribe(self):
        if self.count > 0:
            self.count -= 1
        print(f"Total: {self.count}")

    def show(self):
        print(f"{self.name} has {self.count} subscribers")


abhi = YoutubeChannel("Abhay Inspire")

abhi.subscribe()
abhi.subscribe()
abhi.subscribe()
abhi.subscribe()

abhi.unsubscribe()

abhi.show()
```
