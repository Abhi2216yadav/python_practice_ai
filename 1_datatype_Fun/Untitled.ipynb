```python
revenues = [50, 60 , 70]
expencess = [10, 20, 35]

for i in range(len(revenues)):
    profit = revenues[i] - expencess[i]
    margin = profit * 100 / revenues[i]

    print(f"Margin {i+1} : {margin:.2f}")
```

```python
margins = []

for i in range(len(revenues)):
    profit = revenues[i] - expencess[i]
    margin = profit * 100 / revenues[i]
    margins.append(margin)

margins
    
```

```python
finacial = [
    { "revenue" : 50, "expencess" : 10 },
    { "revenue" : 60, "expencess" : 30 },
    { "revenue" : 55, "expencess" : 35 }
]
finacial
```

```python
for fin in finacial:
    profit = fin["revenue"] - fin["expencess"]
    fin['margin'] = profit * 100 / fin["revenue"]

finacial
```

```python
finacial = {
    "Q1" : { "revenue" : 50, "expencess" : 10 },
    "Q2" : { "revenue" : 60, "expencess" : 30 },
    "Q3" : { "revenue" : -70, "expencess" : 30 },
    "Q4" : { "revenue" : 55, "expencess" : 35 }
}

for quarter, data in finacial.items():
    # print("Quarter : ",quarter)
    # print("Data : ", data)

    if data["revenue"] < 0:
        print(f"Invalid revenue for {quarter}. skipping")
        continue
    margin = (data["revenue"] - data["expencess"]) / data["revenue"] * 100
    print(f"{quarter} : {margin:.2f}% ")
    
```
