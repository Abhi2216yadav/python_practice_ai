```python
def safe_margin(revenue, expencesses):
    try:
        return (revenue - expencesses) * 100 / revenue
    except ZeroDivisionError:
        return "ZeroDivisionError occure in this case"
print(safe_margin(10,3))
print(safe_margin(0, 3)) 
```

### Chai bill calculator

```python
def safe_chai_bill(cups, price_per_cup):
    try:
        if cups < 1:
            return "Cups must be atleast one"
        if price_per_cup < 1:
            return "Cups price must be atleast one rupee"
        return cups * price_per_cup
    except Exception as e:
        return "Please enter numbers only"

safe_chai_bill(2, 15)
safe_chai_bill(32, 1)
# safe_chai_bill(2, "15")
# safe_chai_bill("Five", 15)
```
