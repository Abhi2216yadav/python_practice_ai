```python
conversation = []    # this list IS the chatbot's memory

def chat(user_message):
    # 1. add what the user said to memory
    conversation.append({"role": "user", "parts": [{"text": user_message}]})

    # 2. send the ENTIRE conversation so far
    try:
        response = client.models.generate_content(
            # model="gemini-3.1-pro",
            model="gemini-3.7-flash",
            contents=conversation
        )
        reply = response.text
    except Exception as e:
        return f"Sorry, something went wrong: {e}"

    # 3. add the AI's reply to memory too, so it remembers next time
    conversation.append({"role": "model", "parts": [{"text": reply}]})
    return reply
```

```python
chat("Hi! My name is Abhay. Please remember it.")
```

```python
for message in conversation:
    print(message["role"], "->", message["parts"][0]["text"])
    print()
```

```python
from pydantic import BaseModel
from google.genai import types

class Recipe(BaseModel):
    title : str #schemas
    calories : int
    pre_time_mint: int
response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="Give me a spisy chiken tanduri recipe",
    config=types.GenerateContentConfig(
        response_mime_type="application/json",
        response_schema=Recipe,
    ),  
)
print(response.text)# in json
# recipe = response.parsed #store as a tuple
# recipe.calories
```

### Custumer review analyzer

```python
from pydantic import BaseModel
from google.genai import types

class ReviewAnalyzer(BaseModel):
    sentiment : str #schemas
    rating : int
    pros : list[str]
    cons : list[str]
response = client.models.generate_content(
    model="gemini-3.5-flash",
    contents="Give me the rating of the Sumsung burds 2 pro",
    config=types.GenerateContentConfig(
        response_mime_type="application/json",
        response_schema=ReviewAnalyzer,
    ),  
)
print(response.text)# in json
# recipe = response.parsed #store as a tuple
# recipe.calories
```
